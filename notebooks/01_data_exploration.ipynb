# 01 · Data Exploration

Explore the 100k LAION sample that will be used to fit all three generators. Compute baseline statistics for comparison.

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent / "src"))
from utils import read_fbin
from metrics import twonn_lid, hub_skewness
```

## Load sample

Expected file: `data/laion_sample_100k.fbin` (100 000 × 768 float32).

```python
sample = read_fbin("../data/laion_sample_100k.fbin")
print(f"Shape: {sample.shape}, dtype: {sample.dtype}")
```

## Norm distribution

```python
norms = np.linalg.norm(sample, axis=1)
plt.figure(figsize=(6, 3))
plt.hist(norms, bins=60, edgecolor="none")
plt.xlabel("L2 norm")
plt.ylabel("count")
plt.title("Vector norms — LAION 100k sample")
plt.tight_layout()
plt.savefig("../results/01_norms.png", dpi=120)
plt.show()
```

## Intrinsic dimensionality (TwoNN)

```python
lid = twonn_lid(sample)
print(f"TwoNN LID: {lid:.2f}")
```

## Hub skewness

```python
skew = hub_skewness(sample)
print(f"N_10 skewness: {skew:.3f}")
```

## Covariance spectrum

```python
cov = np.cov(sample.T)
eigvals = np.linalg.eigvalsh(cov)[::-1]
pr = eigvals.sum()**2 / (eigvals**2).sum()
print(f"Participation ratio (PR): {pr:.1f}")
plt.figure(figsize=(6, 3))
plt.plot(np.cumsum(eigvals) / eigvals.sum())
plt.axhline(0.9, color="red", linestyle="--", label="90%")
plt.xlabel("component")
plt.ylabel("cumulative variance")
plt.title("Scree plot — LAION 100k")
plt.legend()
plt.tight_layout()
plt.savefig("../results/01_scree.png", dpi=120)
plt.show()
```
